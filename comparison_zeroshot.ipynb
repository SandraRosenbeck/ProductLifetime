```python
import pandas as pd
from sklearn.metrics import mutual_info_score, normalized_mutual_info_score, confusion_matrix, ConfusionMatrixDisplay, classification_report
```

```python
# Load predictions
df_clip = pd.read_csv("clip_predictions.csv")
df_blip2 = pd.read_csv("blip2_predictions.csv")
# Merge on image file (just to be safe)
df_all = df_clip.merge(df_blip2, on="image_file", how="inner", suffixes=("_clip", "_blip2"))
df_all["true_label"] = df_all["true_label_clip"] 

true_labels = df_all["true_label"].tolist()
clip_preds = df_all["clip_pred"].tolist()
blip2_preds = df_all["blip2_pred"].tolist()

print("Number of rows after merge:", len(df_all))
```

```python
# Mutual information of ground truth and between models
mi_clip = mutual_info_score(true_labels, clip_preds)
mi_blip2 = mutual_info_score(true_labels, blip2_preds)
mi_clip_blip2 = mutual_info_score(clip_preds, blip2_preds)

print("MI clip:", mi_clip)
print("MI blip2:", mi_blip2)
print("MI clip-blip2:", mi_clip_blip2)
```
